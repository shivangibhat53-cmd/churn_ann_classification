```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
##Load the trained model, scaler and ohe
model = load_model('model.h5')

with open('label_encoder_gender.pkl', 'rb') as file:
    label_encoder_gender = pickle.load(file)

with open('onehot_encoder_geo.pkl','rb') as file:
    onehot_encoder_geo = pickle.load(file)

with open('scaler.pkl', 'rb') as file:
    scaler = pickle.load(file)
```

```python
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
input_data
```

```python
geo_encoded = onehot_encoder_geo.transform([[input_data['Geography']]])
geo_encoded_df = pd.DataFrame(geo_encoded,columns=onehot_encoder_geo.get_feature_names_out())
```

```python
input_df = pd.DataFrame([input_data])
```

```python
input_df['Gender'] = label_encoder_gender.transform(input_df['Gender'])
```

```python
input_data = pd.concat([input_df.reset_index(drop = True),geo_encoded_df], axis = 1)
```

```python
input_data = input_data.drop(['Geography'], axis = 1)
```

```python
input_data
```

```python
#Scaling the input data
input_scaled = scaler.transform(input_data)
input_scaled
```

```python
#Predict churn
prediction = model.predict(input_scaled)
prediction
```

```python
prediction_proba = prediction[0][0]
prediction_proba
```

```python
if prediction_proba > 0.5:
    print('The customer is likely to churn')
else:
    print('The customer is not likely to churn')
```
